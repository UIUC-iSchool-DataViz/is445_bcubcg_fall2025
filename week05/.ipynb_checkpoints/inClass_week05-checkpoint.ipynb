```python
import pandas as pd
```

```python
buildings = pd.read_csv("https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/building_inventory.csv",
                       na_values = {'Square Footage':0, 
                                   'Year Acquired':0,
                                   'Year Constructed':0})
```

```python
buildings.head()
```

```python
stats = buildings.groupby('Year Acquired')['Square Footage'].describe()
stats
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', label='Max')
ax.plot(stats['min'], marker='.', label='Min')
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage (max)')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
plt.style.use('ggplot') # setting the default for all plots in the notebook
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', label='Max')
ax.plot(stats['min'], marker='.', label='Min')
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage (max)')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
style = 'ggplot' # Step 1 -- hard code a parameter we want to change

with plt.style.context(style): # just for this plot
    # moved in a block with CMD+]
    fig, ax = plt.subplots(figsize=(10,2))
    ax.plot(stats['max'], marker='.', label='Max')
    ax.plot(stats['min'], marker='.', label='Min')
    ax.set_xlabel('Year Acquired')
    ax.set_ylabel('Square Footage (max)')
    ax.set_yscale('log')
    ax.legend()
    plt.show()
```

```python
#style = 'ggplot' # Step 2 -- function with parameter as input

def make_plot(style):
    with plt.style.context(style): # just for this plot
        # moved in a block with CMD+]
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', label='Max')
        ax.plot(stats['min'], marker='.', label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage (max)')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
# Step 2.5 -- test the function!
make_plot('grayscale')
```

```python
import ipywidgets
```

```python
@ipywidgets.interact(style=plt.style.available) # Step 3 -- a little "hat" i.e. a decorator with parameter as input
def make_plot(style):
    with plt.style.context(style): # just for this plot
        # moved in a block with CMD+]
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', label='Max')
        ax.plot(stats['min'], marker='.', label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage (max)')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
@ipywidgets.interact(x=10)
def f(x):
    print('my value = ', str(x))
```

```python
#!wget https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/michigan_lld.flt
```

```python
#michigan_file = './michigan_lld.flt' # if using wget, local michigan
michigan_file = '/Users/jnaiman/Downloads/michigan_lld.flt'
```

```python
import numpy as np
```

```python
michigan = np.fromfile(michigan_file, dtype='f4').reshape(5365,4201)
# this is bathymetry data in binary float
# we are reshaping based on our knowledge of the shape of this file
```

```python
plt.hist(michigan.flatten())
plt.show()
```

```python
np.min(michigan)
```

```python
michigan[michigan == -9999] = np.nan
```

```python
plt.hist(michigan.flatten())
plt.show()
```

We want an interactive visualization of this data with:
1. choice of colormap
2. the option to take the log of the colormap

```python
# Step 1: colormap -- hard code a value
colormap = 'Reds'
plt.imshow(michigan, cmap=colormap)
plt.colorbar()
plt.show()
```

```python
# Step 2: make a function with the parameter as input
#colormap = 'Reds'
def make_mich_plot(colormap):
    # CMD + ]
    plt.imshow(michigan, cmap=colormap)
    plt.colorbar()
    plt.show()
```

```python
# Step 2.5 - test the function
make_mich_plot('Blues')
```
